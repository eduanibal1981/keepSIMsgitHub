```python
from llmlingua import PromptCompressor

# 1. Initialize the compressor (uses a tiny, fast model)
compressor = PromptCompressor(
    model_name="microsoft/llmlingua-2-bert-base-multilingual-cased",
    use_llmlingua2=True
)

# 2. Paste your massive code or text here
long_prompt = """
[Paste your huge codebase, logs, or prompt here]
"""

# 3. Compress it by 50% (rate=0.5)
results = compressor.compress_prompt(
    long_prompt,
    rate=0.5,
    force_tokens=["def", "class", "return", "{", "}", "if"] # Keeps code syntax intact
)

print("Original Tokens:", results["origin_tokens"])
print("Compressed Tokens:", results["compressed_tokens"])
print("\n--- COMPRESSED PROMPT TO COPY --- \n")
print(results["compressed_prompt"])
```
